```python
upper_limit=(df3['2022 Population'].mean() + 3*df3['2022 Population'].std())
lower_limit=(df3['2022 Population'].mean()- 3*df3['2022 Population'].std())
print('Upper limit:',upper_limit)
print('lower limit:',lower_limit)
```

```python
df3.loc[(df3['2022 Population']>=upper_limit) | (df3['2022 Population']<=lower_limit)]
```

```python
new_df=df3.loc[(df3['2022 Population']<=upper_limit) & (df3['2022 Population']>=lower_limit)]
print("df before removing outliers",len(df3))
print("df after removing outliers",len(new_df))
print("no of outliers",len(df3)-len(new_df))
```

```python
sns.boxplot(df['2022 Population'])
```

```python
sns.boxplot(new_df['2022 Population'])
```

```python
new_df=df3.copy()
```

```python
new_df.loc[(new_df['2022 Population']>= upper_limit),'2022 Population']=upper_limit
new_df.loc[(new_df['2022 Population']<=lower_limit),'2022 Population']=lower_limit
```

# iqr method


```python
q1=df['2022 Population'].quantile(0.25)
q3=df['2022 Population'].quantile(0.75)
iqr=q3-q1
print(q1,q3,iqr)
```

```python
upper_limit=q3+(1.5*iqr)
lower_limit=q1-(1.5*iqr)
print(upper_limit,lower_limit)
```

```python
df3.loc[(df3['2022 Population']>=upper_limit) | (df3['2022 Population']<=lower_limit)]
```

```python
new_df=df3.loc[(df3['2022 Population']<=upper_limit) & (df3['2022 Population']>=lower_limit)]
print("df before removing outliers",len(df3))
print("df after removing outliers",len(new_df))
print("no of outliers",len(df3)-len(new_df))
```

```python
sns.boxplot(new_df['2022 Population'])
```

```python
new_df=df3.copy()
```

```python
new_df.loc[(new_df['2022 Population']>= upper_limit),'2022 Population']=upper_limit
new_df.loc[(new_df['2022 Population']<=lower_limit),'2022 Population']=lower_limit
```

## percentile method


```python
upper_limit=df3['2022 Population'].quantile(0.99)
lower_limit=df3['2022 Population'].quantile(0.01)
print(upper_limit,lower_limit)
```

```python
df3.loc[(df3['2022 Population']>=upper_limit) | (df3['2022 Population']<=lower_limit)]
```

```python
new_df=df3.loc[(df3['2022 Population']<=upper_limit) & (df3['2022 Population']>=lower_limit)]
print("df before removing outliers",len(df3))
print("df after removing outliers",len(new_df))
print("no of outliers",len(df3)-len(new_df))
```

```python
sns.boxplot(new_df['2022 Population'])
```

```python
## capping
new_df=df3.copy()
new_df.loc[(new_df['2022 Population']>= upper_limit),'2022 Population']=upper_limit
new_df.loc[(new_df['2022 Population']<=lower_limit),'2022 Population']=lower_limit
sns.boxplot(new_df['2022 Population'])
```
